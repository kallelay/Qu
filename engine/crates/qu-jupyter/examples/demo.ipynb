# Qu in a Jupyter notebook

Run these cells in order (Shift+Enter). Each one exercises something the kernel does: persistent state across cells, live `print` output, inline figures, and error recovery.

```qu
x = 5
```

```qu
print("x + 1 = ", x + 1)
```

```qu
y = x * 2;  # trailing ; suppresses the echo
```

```qu
y  # still here from the cell above -- persistent session state
```

```qu
t = 0 to 2*pi step 0.05
plot(t, sin(t))
xlabel("t")
ylabel("sin(t)")
```
